```python
#parsing html
def extract_cao_courses(url, year, default_level=None):
    
    response = requests.get(url, headers=headers)
    response.raise_for_status()
    
    soup = BeautifulSoup(response.text, "html.parser")
    
    records = []
    
    # Each institution's courses are contained in a <pre> block
    for pre in soup.find_all("pre"):
        
        text = pre.get_text("\n", strip=False)
        lines = text.splitlines()
        
        # Find the institution name from the bold header
        bold = pre.find("b")
        
        if not bold:
            continue
        
        header = bold.get_text(" ", strip=True)
        
        # Only process blocks containing a course code
        course_lines = [
            line.strip()
            for line in lines
            if re.match(r"^[A-Z]{2}\d{3}\s+", line.strip())
        ]
        
        if not course_lines:
            continue
        
        # Institution is the line after "RND 1" in the header
        header_lines = [x.strip() for x in header.splitlines() if x.strip()]
        
        institution = None
        
        for i, line in enumerate(header_lines):
            if "RND" in line.upper() and i + 1 < len(header_lines):
                institution = header_lines[i + 1]
                break
        
        if institution is None:
            continue
        
        for line in course_lines:
            
            # Course code
            match = re.match(r"^([A-Z]{2}\d{3})\s+(.*)$", line)
            
            if not match:
                continue
            
            course_code = match.group(1)
            remainder = match.group(2).strip()
            
            # Points are at the end of the line.
            # They may be:
            # 241
            # #862
            # 409*
            # #625*
            # AQA
            # or blank
            points_match = re.search(r"(#?\d+\*?|AQA)\s*$", remainder)
            
            if points_match:
                points_raw = points_match.group(1)
                course_title = remainder[:points_match.start()].strip()
            else:
                points_raw = ""
                course_title = remainder.strip()
            
            # AQA flag
            aqa = points_raw == "AQA"
            
            # Remove # and * from numeric points
            if points_raw and points_raw != "AQA":
                points = re.sub(r"[#*]", "", points_raw)
                points = int(points)
            else:
                points = None
            
            # R1 Points: numeric score, or blank for AQA/no points
            r1_points = points
            
            records.append({
                "Course Title": course_title,
                "Course Code": course_code,
                "R1 Points": r1_points,
                "HE Institution": institution,
                "AQA": aqa,
                "Year": year,
                "Points": points,
                "Course Level": default_level
            })
    
    return pd.DataFrame(records)
```

```python
#extracting both pages
df_l8 = extract_cao_courses(
    url_l8,
    year=2026,
    default_level=8
)

df_l67 = extract_cao_courses(
    url_l67,
    year=2026,
    default_level=None
)

print("Level 8 courses:", len(df_l8))
print("Level 6/7 courses:", len(df_l67))
```
