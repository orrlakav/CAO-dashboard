# Creating larger dataset

```python
all_cao_data = pd.concat(
    [
        data2021_clean,
        data2022_clean,
        data2023_clean,
        data2024_clean,
        data2025_clean
    ],
    ignore_index=True
)

all_cao_data.head()
```

```python
#check cols
print(all_cao_data.columns.tolist())
```

```python
#checking to see if any courses had 0 points as the requirment 
all_cao_data[all_cao_data["R1 Points"] == 0]
```

```python
#changing all AQA values to 0 points- information will still be retained by AQA column
all_cao_data["Points"] =  pd.to_numeric(
    all_cao_data["R1 Points"],
    errors="coerce"
).astype("Int64")

all_cao_data.loc[all_cao_data["AQA"], "Points"] = 0
#show
all_cao_data.head(20)
```

```python
#changing catergory column title and HEI
all_cao_data = all_cao_data.rename(
    columns={"CATEGORY (ISCED Description)": "Category", "HEI":"HE Institution"}
)
```

```python
all_cao_data.head()
```

```python
#fixing duplicate categories...
#working out which category has most years of data
year_columns = [2021, 2022, 2023, 2024, 2025]

category_choice = (
    all_cao_data
    .groupby(["Course Code", "Category"], dropna=False)
    ["Points"]
    .apply(lambda x: x.notna().sum())
    .reset_index(name="Years_with_data")
)

category_choice.head()
```

```python
#choose category with most years of data
category_choice = (
    category_choice
    .sort_values(
        ["Course Code", "Years_with_data"],
        ascending=[True, False]
    )
    .drop_duplicates("Course Code")
)
```

```python
#lookup 
category_lookup = category_choice.set_index("Course Code")["Category"]
```

```python
#replacing category throughout 
all_cao_data["Category"] = (
    all_cao_data["Course Code"]
    .map(category_lookup)
)
```

```python
#checking one course
all_cao_data[
    all_cao_data["Course Code"] == "CK412"
][
    ["Course Code", "Course Title", "Category", "Year", "Points"]
].sort_values("Year")
```

```python
#pivot on year for trend graphs- using course code as identifier
course_history = all_cao_data.pivot_table(
    index=["Course Code",
        "HE Institution",
        "Course Level",
        "Category"
          ],
    columns="Year",
    values="Points",
    aggfunc="first"
).reset_index()

course_history.head()
```

```python
# get course titles back (issue with duplicate titles with slight changes)
# Get the 2025 course titles
course_titles = data2025_clean[
    ["Course Code", "Course Title"]
].drop_duplicates("Course Code")

# Add the 2025 course titles to the historical data
course_history = course_history.merge(
    course_titles,
    on="Course Code",
    how="inner"    #changed to inner as only want 2025 list... others may be obsolete courses
)

# Move Course Title beside Course Code
course_history = course_history[
    [
        "Course Code",
        "Course Title",
        "HE Institution",
        "Course Level",
        "Category",
        2021,
        2022,
        2023,
        2024,
        2025
    ]
]

course_history.head()
```
